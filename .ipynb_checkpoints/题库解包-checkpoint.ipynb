```python
import os, shutil, time, hashlib
from datetime import datetime

watch_dir = r"D:\KSWJJ\61000001"
backup_dir = r"F:\All_Proj"

if not os.path.exists(backup_dir):
    os.makedirs(backup_dir)

set_count = 1
proj_expected = {"proj1", "proj2", "proj3"}
current_set_dir = None
seen_hash = {}

def get_folder_hash(path):
    md5 = hashlib.md5()
    for root, _, files in os.walk(path):
        for f in files:
            file_path = os.path.join(root, f)
            with open(file_path, 'rb') as fp:
                while chunk := fp.read(4096):
                    md5.update(chunk)
    return md5.hexdigest()

def copy_proj():
    global current_set_dir
    for root, dirs, _ in os.walk(watch_dir):
        for d in dirs:
            if d.startswith("proj"):
                src = os.path.join(root, d)
                folder_hash = get_folder_hash(src)

                # 如果内容没变化 → 跳过
                if seen_hash.get(d) == folder_hash:
                    continue
                seen_hash[d] = folder_hash

                # 初始化 _temp 目录
                if current_set_dir is None:
                    current_set_dir = os.path.join(backup_dir, "_temp")
                    if not os.path.exists(current_set_dir):
                        os.makedirs(current_set_dir)

                dst = os.path.join(current_set_dir, d)
                if os.path.exists(dst):
                    shutil.rmtree(dst)
                shutil.copytree(src, dst)
                print(f"✅ 检测到更新并备份: {dst}")

print("📡 开始监控，按 Enter 手动确认一套完成")

while True:
    copy_proj()
    if input().strip() == "":   # 手动回车确认一套
        if current_set_dir:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            final_dir = os.path.join(backup_dir, f"Set_{set_count}_{timestamp}")
            shutil.move(current_set_dir, final_dir)
            print(f"🎉 第 {set_count} 套题完成 → {final_dir}")
            set_count += 1
            current_set_dir = None
        else:
            print("⚠️ 当前没有检测到新的 proj，回车无效")
```

```python
import os, shutil, hashlib, time
from datetime import datetime

# === 配置路径 ===
watch_dir = r"D:\KSWJJ\61000001"
backup_dir = r"F:\All_Proj"

if not os.path.exists(backup_dir):
    os.makedirs(backup_dir)

seen_hash = set()
set_count = 1
proj_expected = {"proj1", "proj2", "proj3"}
current_group = {}

# 计算文件夹哈希
def get_folder_hash(path):
    md5 = hashlib.md5()
    total_size = 0
    for root, _, files in os.walk(path):
        for f in sorted(files):   # 排序避免不同顺序导致哈希不同
            file_path = os.path.join(root, f)
            total_size += os.path.getsize(file_path)
            with open(file_path, 'rb') as fp:
                while chunk := fp.read(4096):
                    md5.update(chunk)
    return f"{total_size}_{md5.hexdigest()}"

# 备份
def backup_proj():
    global set_count, current_group
    for root, dirs, _ in os.walk(watch_dir):
        for d in dirs:
            if d.startswith("proj"):
                src = os.path.join(root, d)
                folder_hash = get_folder_hash(src)

                # ✅ 去重保护
                if folder_hash in seen_hash:
                    continue

                dst_group = os.path.join(backup_dir, f"Set_{set_count}")
                if not os.path.exists(dst_group):
                    os.makedirs(dst_group)

                dst = os.path.join(dst_group, d)
                shutil.copytree(src, dst)
                seen_hash.add(folder_hash)
                current_group[d] = dst
                print(f"✅ 已备份: {dst}")

    # 检查是否三份都齐了
    if proj_expected.issubset(current_group.keys()):
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        final_name = os.path.join(backup_dir, f"Set_{set_count}_{timestamp}")
        os.rename(os.path.join(backup_dir, f"Set_{set_count}"), final_name)
        print(f"🎉 第 {set_count} 套题完成 → {final_name}")
        set_count += 1
        current_group.clear()

print("📡 开始监控（自动去重保护）...")
while True:
    backup_proj()
    time.sleep(1)
```

```python
import os, shutil, hashlib, time
from datetime import datetime

# === 配置路径 ===
watch_dir = r"D:\KSWJJ\61000001"
backup_dir = r"F:\All_Proj"

if not os.path.exists(backup_dir):
    os.makedirs(backup_dir)

seen_hash = set()
set_count = 1
proj_expected = {"proj1", "proj2", "proj3"}
current_group = {}

# ✅ 忽略的临时文件扩展名
ignore_ext = {".tmp", ".log", ".bak"}

def get_folder_hash(path):
    md5 = hashlib.md5()
    total_size = 0
    for root, _, files in os.walk(path):
        for f in sorted(files):
            file_path = os.path.join(root, f)
            if os.path.splitext(f)[1].lower() in ignore_ext:
                continue
            try:
                total_size += os.path.getsize(file_path)
                with open(file_path, 'rb') as fp:
                    while chunk := fp.read(4096):
                        md5.update(chunk)
            except FileNotFoundError:
                print(f"⚠️ 跳过丢失文件: {file_path}")
                continue
    return f"{total_size}_{md5.hexdigest()}"

def backup_proj():
    global set_count, current_group
    found = []

    for root, dirs, _ in os.walk(watch_dir):
        for d in dirs:
            if d.startswith("proj"):
                src = os.path.join(root, d)
                folder_hash = get_folder_hash(src)

                # ✅ 去重保护
                if folder_hash in seen_hash:
                    continue

                dst_group = os.path.join(backup_dir, f"Set_{set_count}_{datetime.now().strftime('%Y%m%d_%H%M%S')}")
                if not os.path.exists(dst_group):
                    os.makedirs(dst_group)

                dst = os.path.join(dst_group, d)
                shutil.copytree(src, dst)
                seen_hash.add(folder_hash)
                current_group[d] = dst
                print(f"✅ 已备份: {dst}")

    # ✅ 检查是否三份都齐了
    if proj_expected.issubset(current_group.keys()):
        print(f"🎉 第 {set_count} 套题完成 → {os.path.dirname(list(current_group.values())[0])}")
        set_count += 1
        current_group = {}

print("📡 开始监控（自动去重 + 临时文件过滤）...")
while True:
    backup_proj()
    time.sleep(1)
```

```python
import os

backup_dir = r"F:\All_Proj"

def get_folder_size(path):
    total_size = 0
    for root, _, files in os.walk(path):
        for f in files:
            file_path = os.path.join(root, f)
            total_size += os.path.getsize(file_path)
    return total_size

# 统计文件夹大小
sizes = {}
duplicates = {}

for folder in os.listdir(backup_dir):
    full_path = os.path.join(backup_dir, folder)
    if os.path.isdir(full_path):
        size = get_folder_size(full_path)
        if size in sizes:
            duplicates.setdefault(size, []).append(full_path)
            duplicates[size].append(sizes[size])
        else:
            sizes[size] = full_path

# 输出结果
if duplicates:
    print("✅ 检测到大小完全相同的文件夹：")
    for size, paths in duplicates.items():
        unique_paths = list(set(paths))
        print(f"\n大小: {size} 字节")
        for p in unique_paths:
            print("  ", p)
else:
    print("没有检测到大小相同的文件夹")
```

```python
import os, shutil, hashlib
from pathlib import Path

backup_dir = r"F:\All_Proj"

def get_folder_hash(path):
    md5 = hashlib.md5()
    for root, _, files in os.walk(path):
        for f in sorted(files):
            file_path = os.path.join(root, f)
            with open(file_path, 'rb') as fp:
                while chunk := fp.read(4096):
                    md5.update(chunk)
    return md5.hexdigest()

# 收集所有 Set 文件夹
folders = [os.path.join(backup_dir, d) for d in os.listdir(backup_dir) if os.path.isdir(os.path.join(backup_dir, d))]

# 按创建时间排序
folders.sort(key=lambda p: os.path.getctime(p))

# 去重
seen = set()
unique_folders = []
for f in folders:
    h = get_folder_hash(f)
    if h not in seen:
        seen.add(h)
        unique_folders.append(f)
    else:
        print(f"⚠️ 删除重复: {f}")
        shutil.rmtree(f)

# 重新命名
for idx, folder in enumerate(unique_folders, start=1):
    new_name = os.path.join(backup_dir, f"Set_{idx}")
    if folder != new_name:
        print(f"重命名: {folder} → {new_name}")
        os.rename(folder, new_name)

print("✅ 去重+重命名完成")
```

```python
import os

backup_dir = r"F:\All_Proj"
required = {"proj1", "proj2", "proj3"}

incomplete_sets = []

for folder in os.listdir(backup_dir):
    full_path = os.path.join(backup_dir, folder)
    if not os.path.isdir(full_path):
        continue
    
    subfolders = {d for d in os.listdir(full_path) if os.path.isdir(os.path.join(full_path, d))}
    missing = required - subfolders
    if missing:
        incomplete_sets.append((folder, missing))

if incomplete_sets:
    print("⚠️ 检测到 project 不完整的文件夹：")
    for set_name, missing in incomplete_sets:
        print(f"  {set_name} → 缺少: {', '.join(sorted(missing))}")
else:
    print("✅ 所有文件夹都包含 proj1/proj2/proj3")
```

```python
import os
from PIL import Image
from pygments import highlight
from pygments.lexers import CppLexer
from pygments.formatters import ImageFormatter

cpp_dir = r"F:\All_Proj"
save_dir = r"F:\cpp_images"

if not os.path.exists(save_dir):
    os.makedirs(save_dir)

for root, _, files in os.walk(cpp_dir):
    for file in files:
        if file.endswith(".cpp"):
            file_path = os.path.join(root, file)

            # ✅ 尝试多种编码
            for enc in ["utf-8", "gbk", "latin-1"]:
                try:
                    with open(file_path, "r", encoding=enc) as f:
                        code = f.read()
                    break
                except UnicodeDecodeError:
                    continue

            # ✅ 渲染成图片
            img_data = highlight(code, CppLexer(), ImageFormatter(font_size=18, line_numbers=True))

            # 保持目录结构
            rel_path = os.path.relpath(file_path, cpp_dir)
            img_path = os.path.join(save_dir, rel_path).replace(".cpp", ".png")

            os.makedirs(os.path.dirname(img_path), exist_ok=True)
            with open(img_path, "wb") as img_file:
                img_file.write(img_data)

            print(f"✅ 转换完成: {img_path}")
```

```python
import os
from pygments import highlight
from pygments.lexers import CppLexer
from pygments.formatters import ImageFormatter

cpp_dir = r"F:\All_Proj"
save_dir = r"F:\cpp_images"

if not os.path.exists(save_dir):
    os.makedirs(save_dir)

for root, _, files in os.walk(cpp_dir):
    for file in files:
        if file.endswith(".cpp"):
            file_path = os.path.join(root, file)

            # 尝试多种编码
            for enc in ["utf-8", "gbk", "latin-1"]:
                try:
                    with open(file_path, "r", encoding=enc) as f:
                        code = f.read()
                    break
                except UnicodeDecodeError:
                    continue

            # ✅ VSCode 风格 + 中文字体
            font_path = r"C:\Windows\Fonts\msyh.ttc"  # 微软雅黑路径
            formatter = ImageFormatter(
                font_name=font_path,
                font_size=18,
                line_numbers=True,
                style="monokai"
            )
            img_data = highlight(code, CppLexer(), formatter)

            # 保持目录结构
            rel_path = os.path.relpath(file_path, cpp_dir)
            img_path = os.path.join(save_dir, rel_path).replace(".cpp", ".png")
            os.makedirs(os.path.dirname(img_path), exist_ok=True)

            with open(img_path, "wb") as img_file:
                img_file.write(img_data)

            print(f"✅ 转换完成: {img_path}")
```
