### Deep learning model configurations


*   Feed-forward neural network: set architecture parameter to 'nn'
*   Convolutional neural network: architecture = 'cnn'
*   Long short-term memory model: architecture = 'lstm'
*   Long short-term memory with convolutional neural network layer: architecture = 'lstm_cnn'

For each model, please specificy the following hyperparameters:

1.   input_shape (in this project, the input_shape is (12,) for CumRet1 ~ CumRet12)
2.   filters: number of filters in CNN
3.  kernel_size: the size of kernel used in CNN
4. lstm_num: number of nodes in LSTM
5. hidden_dims: number of hidden nodes in NN
6. dropout: dropout rate





```python
def my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout):
    if architecture == 'nn':
        model = Sequential()
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'cnn':
        model = Sequential()
        model.add(Reshape((input_shape, 1),input_shape=(input_shape,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        model.add(MaxPooling1D())
        model.add(Flatten())
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'lstm':
        model = Sequential()
        model.add(Reshape((input_shape, 1), input_shape=(input_shape,)))
        model.add(LSTM(hidden_dims))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'cnn_lstm':
        model = Sequential()
        model.add(Reshape((input_shape, 1),input_shape=(input_shape,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        model.add(MaxPooling1D())
        model.add(LSTM(lstm_num))
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'lstm_cnn':
        model = Sequential()
        model.add(Reshape((input_shape, 1), input_shape=(input_shape,)))
        model.add(LSTM(hidden_dims))
        model.add(Reshape((hidden_dims, 1),input_shape=(hidden_dims,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        model.add(MaxPooling1D())
        model.add(Flatten())
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    return model
```

```python
# plot the loss function with respect to each epoch
def my_plot(history,model_path):
    # Plot training & validation loss values
    fig = plt.figure()
    plt.plot(history.history['mae'])
    plt.plot(history.history['val_mae'])
    plt.title('Model MAE Comparison')
    plt.ylabel('MAE Measure')
    plt.xlabel('Epoch')
    plt.legend(['Train', 'Validate'], loc='upper left')
    fig.savefig(model_path + '.png')
    plt.close()
    # plt.show()
```

## Optimization Period Fits
The following section is to use data from 192701 to 196306 to find the optimal/best model configuration. Specifically,
for models corresponding to each combination of
> arch (FNN, CNN, LSTM, and CNNLSTM),

> lossfun (MAE, MSE),

> weight (EW, EWPM, EWPMVW),

> retvar (RetRF, RetRFStd, RetRFPctl, RetRFNorm),

and fit the model 30 times using only observations from months (Month column in file) between 192701-196306 corresponding to even years from even months and odd years from odd months.  The fits should be based on CumRet1,..., CumRet12.

For each configurartion, the results for the i-th run are saved in
basedir/MLForecasts/OptimizationPeriod/arch_lossfun_weight_retvar/vi/Forecast.csv

Once all of 30 runs are finished, the average over 30 Forecast.csv will be calculated and saved in
basedir/MLForecasts/OptimizationPeriod/arch_lossfun_weight_retvar/mean/Forecast.csv

```python
even_month = [2,4,6,8,10,12]
odd_month = [1,3,5,7,9,11]
year = [w for w in range(1927,1964,1)]
even_year = [w for w in year if w % 2 == 0]
odd_year = [w for w in year if w % 2 == 1]
```

```python
MinMonth,MaxMonth = 192701,196306
features = ['CumRet1','CumRet2','CumRet3','CumRet4','CumRet5','CumRet6',
            'CumRet7','CumRet8','CumRet9','CumRet10','CumRet11','CumRet12']
Observations = data[(data['Month']>=MinMonth) & (data['Month']<=MaxMonth)]
Observations['Y'] = Observations.apply(lambda x: int(str(x['Month'])[0:4]), axis = 1)
Observations['M'] = Observations.apply(lambda x: int(str(x['Month'])[4:6]), axis = 1)
```

```python
FitObservationsFeaturesCSV = Observations[((Observations['Y'].isin(even_year)) & (Observations['M'].isin(even_month))) | ((Observations['Y'].isin(odd_year))&(Observations['M'].isin(odd_month)))]
PredictionFeaturesCSV      = Observations[((Observations['Y'].isin(even_year)) & (Observations['M'].isin(odd_month)))  | ((Observations['Y'].isin(odd_year))&(Observations['M'].isin(even_month)))]
TestingX = PredictionFeaturesCSV[features].values
```

```python
architectureList = ['nn','cnn','lstm','cnn_lstm']
lossfunctionList = ['mae','mse']
DepVarList = ['RetRF', 'RetRFNorm', 'RetRFStd', 'RetRFPctl']
WeightVarList =  ['EW', 'EWPM', 'EWPMVW']

ModelList = []
for a in architectureList:
    for l in lossfunctionList:
        for d in DepVarList:
            for w in WeightVarList:
                ModelList.append([a, l, w, d])
```

```python
try:
    os.stat(os.path.join(basedir,'MLForecasts','OptimizationPeriod'))
except:
    os.mkdir(os.path.join(basedir,'MLForecasts','OptimizationPeriod'))
for i in range(0,len(ModelList)):
    m = ModelList[i]
    print(m)
    architecture,lossfunction,WeightVar,DepVar = m[0],m[1],m[2],m[3]
    print(architecture, lossfunction, WeightVar, DepVar)
    save_path_head = os.path.join(path,'ChartingByMachines','MLForecasts','OptimizationPeriod'
                                  ,'_'.join([architecture.upper(),lossfunction.upper(),DepVar,WeightVar]))
    try:
        os.stat(save_path_head)
    except:
        os.mkdir(save_path_head)
    N = 30 # we will run the model 30 times and then average them as the final result
    for n in range(N):
      Training,Validating = train_test_split(FitObservationsFeaturesCSV,test_size=0.3)
      TrainingX,Trainingy,TrainingW = Training[features].values,Training[DepVar].values,Training[WeightVar].values
      ValidatingX,Validatingy,ValidatingW = Validating[features].values,Validating[DepVar].values,Validating[WeightVar].values
      period = '192701-196306'
      FolderName = period
      v = 'v'+str(n+1)
      print('period:%s, %s' % (period, v))
      try:
          os.stat(os.path.join(save_path_head,FolderName))
      except:
          os.mkdir(os.path.join(save_path_head,FolderName))
      try:
          os.stat(os.path.join(save_path_head,FolderName,v))
      except:
          os.mkdir(os.path.join(save_path_head,FolderName,v))

      tf.keras.backend.clear_session()
      # early stopping
      es = EarlyStopping(monitor='val_mae', mode='min', verbose=0, patience = 5)
      mc = ModelCheckpoint(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'),
                            monitor='val_mae', mode='min', verbose=0, save_best_only=True)
      input_shape = TrainingX.shape[1]
      model = my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout)
      model.compile(loss=lossfunction, optimizer='adam', metrics=['mae'])
      history = model.fit(TrainingX, Trainingy,
                      sample_weight = TrainingW,
                      batch_size=batch_size,
                      epochs=epochs,
                      validation_data=(ValidatingX, Validatingy),
                      verbose = 0,
                      callbacks = [es, mc])

      saved_model = load_model(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'))

      if lossfunction == 'mse':
          loss = pd.DataFrame(columns=['mse','val_mse','mae','val_mae',])
          loss['mse'] = history.history['loss']
          loss['val_mse'] = history.history['val_loss']
          loss['mae'] = history.history['mae']
          loss['val_mae'] = history.history['val_mae']
      else:
          loss = pd.DataFrame(columns=['mae','val_mae','mae','val_mae',])
          loss['mae'] = history.history['loss']
          loss['val_mae'] = history.history['val_loss']
          loss['mae'] = history.history['mae']
          loss['val_mae'] = history.history['val_mae']

      if 'PredictedValue' in PredictionFeaturesCSV.columns:
          PredictionFeaturesCSV = PredictionFeaturesCSV.drop(columns=['PredictedValue'])

      y = saved_model.predict(TestingX,batch_size=batch_size)

      PredictionFeaturesCSV = PredictionFeaturesCSV.assign(PredictedValue = y)
      PredictionFeaturesCSV[['Month','permno',DepVar,'PredictedValue']].to_csv(os.path.join(save_path_head,FolderName,v,'Forecast.csv'),index=False)
      loss.to_csv(os.path.join(save_path_head,FolderName,v,'LossFunction.csv'),index=False)
```

```python
%%time
modelperiod = ['192701-196306']
N = 30
for i in range(0,len(ModelList)):
    m = ModelList[i]
    architecture,lossfunction,WeightVar,DepVar = m[0],m[1],m[2],m[3]
    print(architecture, lossfunction, WeightVar, DepVar)
    save_path_head = os.path.join(path,'ChartingByMachines','MLForecasts','OptimizationPeriod'
                                  ,'_'.join([architecture.upper(),lossfunction.upper(),DepVar,WeightVar]))
    l = [str(n+1) for n in range(N)]
    for period in modelperiod:
        FolderName = period
        print(period)
        try:
            os.stat(os.path.join(save_path_head,FolderName,'mean'))
        except:
            os.mkdir(os.path.join(save_path_head,FolderName,'mean'))

        d = data[['Month','permno']]
        d = d.drop_duplicates(subset=['Month','permno'])
        for n in range(N):
            v = 'v'+str(n+1)
            d1 = pd.read_csv(os.path.join(save_path_head,FolderName,v,'Forecast.csv'))
            d1 = d1.drop(columns=DepVar)
            d1 = d1.rename(columns={'PredictedValue':str(n+1)})
            d1 = d1.drop_duplicates(subset=['Month','permno'])
            d = d.merge(d1,on=['Month','permno'])
            print(d.shape)
        d['PredictedValue'] = d[l].mean(axis=1)
        d[['Month','permno','PredictedValue']].to_csv(os.path.join(save_path_head,FolderName,'mean','Forecast.csv'), index=False)
```

```python
try:
    os.stat(os.path.join(basedir,'MLForecasts','Main'))
except:
    os.mkdir(os.path.join(basedir,'MLForecasts','Main'))
for i in range(0,len(ModelList)):
    m = ModelList[i]
    architecture,lossfunction,WeightVar,DepVar = m[0],m[1],m[2],m[3]
    print(architecture, lossfunction, WeightVar, DepVar)
    save_path_head = os.path.join(path,'ChartingByMachines','MLForecasts','Main'
                                  ,'_'.join([architecture.upper(),lossfunction.upper(),DepVar,WeightVar]))
    try:
        os.stat(save_path_head)
    except:
        os.mkdir(save_path_head)
    N = 30 # we will run the model 30 times and then average them as the final result
    for period in modelperiod:
        MinMonth = int(period.split('-')[0])
        MaxMonth = int(period.split('-')[1])
        FitObservationsFeaturesCSV = data[(data['Month']>=MinMonth)
                                          & (data['Month']<=MaxMonth)][features+[DepVar] + [WeightVar]]
        TestingX = data[features].values
        for n in range(N):
            Training,Validating = train_test_split(FitObservationsFeaturesCSV,test_size=0.3)
            TrainingX,Trainingy,TrainingW = Training[features].values,Training[DepVar].values,Training[WeightVar].values
            ValidatingX,Validatingy,ValidatingW = Validating[features].values,Validating[DepVar].values,Validating[WeightVar].values
            FolderName = period
            v = 'v'+str(n+1)
            print('period:%s, %s' % (period, v))
            try:
                os.stat(os.path.join(save_path_head,FolderName))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName))
            try:
                os.stat(os.path.join(save_path_head,FolderName,v))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName,v))

            tf.keras.backend.clear_session()
            # early stopping
            es = EarlyStopping(monitor='val_mae', mode='min', verbose=0, patience = 5)
            mc = ModelCheckpoint(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'),
                                 monitor='val_mae', mode='min', verbose=0, save_best_only=True)
            input_shape = TrainingX.shape[1]
            model = my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout)
            model.compile(loss=lossfunction, optimizer='adam', metrics=['mae'])
            history = model.fit(TrainingX, Trainingy,
                            sample_weight = TrainingW,
                            batch_size=batch_size,
                            epochs=epochs,
                            validation_data=(ValidatingX, Validatingy),
                            verbose = 0,
                            callbacks = [es, mc])

            saved_model = load_model(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'))

            if lossfunction == 'mse':
                loss = pd.DataFrame(columns=['mse','val_mse','mae','val_mae',])
                loss['mse'] = history.history['loss']
                loss['val_mse'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']
            else:
                loss = pd.DataFrame(columns=['mae','val_mae','mae','val_mae',])
                loss['mae'] = history.history['loss']
                loss['val_mae'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']

            if 'PredictedValue' in data.columns:
                data = data.drop(columns=['PredictedValue'])

            y = saved_model.predict(TestingX,batch_size=batch_size)

            data = data.assign(PredictedValue = y)
            data[['Month','permno',DepVar,'PredictedValue']].to_csv(os.path.join(save_path_head,FolderName,v,'Forecast.csv'),index=False)
            loss.to_csv(os.path.join(save_path_head,FolderName,v,'LossFunction.csv'),index=False)
```

Similarly, for each model, we will run 30 times and each of the i-th results will be saved in

basedir/MLForecasts/Baselines/architecture_lossfunction_weight_depvar/vi/Forecast.csv

basedir/MLForecasts/Baselines/architecture_lossfunction_weight_depvar/mean/Forecast.csv

```python
try:
    os.stat(os.path.join(basedir,'MLForecasts','Baselines'))
except:
    os.mkdir(os.path.join(basedir,'MLForecasts','Baselines'))
for i in range(0,len(ModelList)):
    m = ModelList[i]
    architecture,lossfunction,WeightVar,DepVar = m[0],m[1],m[2],m[3]
    print(architecture, lossfunction, WeightVar, DepVar)
    save_path_head = os.path.join(path,'ChartingByMachines','MLForecasts','Baselines'
                                  ,'_'.join([architecture.upper(),lossfunction.upper(),DepVar,WeightVar]))
    try:
        os.stat(save_path_head)
    except:
        os.mkdir(save_path_head)
    N = 30 # we will run the model 30 times and then average them as the final result
    for period in modelperiod:
        MinMonth = int(period.split('-')[0])
        MaxMonth = int(period.split('-')[1])
        FitObservationsFeaturesCSV = data[(data['Month']>=MinMonth)
                                          & (data['Month']<=MaxMonth)][features+[DepVar] + [WeightVar]]
        TestingX = data[features].values
        for n in range(N):
            Training,Validating = train_test_split(FitObservationsFeaturesCSV,test_size=0.3)
            TrainingX,Trainingy,TrainingW = Training[features].values,Training[DepVar].values,Training[WeightVar].values
            ValidatingX,Validatingy,ValidatingW = Validating[features].values,Validating[DepVar].values,Validating[WeightVar].values
            FolderName = period
            v = 'v'+str(n+1)
            print('period:%s, %s' % (period, v))
            try:
                os.stat(os.path.join(save_path_head,FolderName))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName))
            try:
                os.stat(os.path.join(save_path_head,FolderName,v))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName,v))

            tf.keras.backend.clear_session()
            # early stopping
            es = EarlyStopping(monitor='val_mae', mode='min', verbose=0, patience = 5)
            mc = ModelCheckpoint(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'),
                                 monitor='val_mae', mode='min', verbose=0, save_best_only=True)
            input_shape = TrainingX.shape[1]
            model = my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout)
            model.compile(loss=lossfunction, optimizer='adam', metrics=['mae'])
            history = model.fit(TrainingX, Trainingy,
                            sample_weight = TrainingW,
                            batch_size=batch_size,
                            epochs=epochs,
                            validation_data=(ValidatingX, Validatingy),
                            verbose = 0,
                            callbacks = [es, mc])

            saved_model = load_model(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'))

            if lossfunction == 'mse':
                loss = pd.DataFrame(columns=['mse','val_mse','mae','val_mae',])
                loss['mse'] = history.history['loss']
                loss['val_mse'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']
            else:
                loss = pd.DataFrame(columns=['mae','val_mae','mae','val_mae',])
                loss['mae'] = history.history['loss']
                loss['val_mae'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']

            if 'PredictedValue' in data.columns:
                data = data.drop(columns=['PredictedValue'])

            y = saved_model.predict(TestingX,batch_size=batch_size)

            data = data.assign(PredictedValue = y)
            data[['Month','permno',DepVar,'PredictedValue']].to_csv(os.path.join(save_path_head,FolderName,v,'Forecast.csv'),index=False)
            loss.to_csv(os.path.join(save_path_head,FolderName,v,'LossFunction.csv'),index=False)
```

## Forecasts Based on Momentum and Reversal

Instead of using CumRet1 ~ CumRet12, in this section we will use the Mom and Rev to train the model and build the forecasts.

We run the model again 30 runs and at each run i, the results are saved in
basedir/MLForecasts/MomRev/CNNLSTM_MSE_EWPM_RetRFNorm/vi/Forecast.csv
and the average results are then saved in
basedir/MLForecasts/MomRev/CNNLSTM_MSE_EWPM_RetRFNorm/mean/Forecast.csv

```python
# input features
features = ['Mom','Rev']
# fitting subperiods
modelperiod = ['192701-196306', '192701-197412', '192701-198412', '192701-199412', '192701-200412', '192701-201412']
# model configuration
ModelList = [
    ['cnn_lstm', 'mse', 'EWPM','RetRFNorm']
    ]
```

```python
# set parameters:
batch_size = 32768
filters = 64
kernel_size = 2
lstm_num = 32
hidden_dims = 64
# epochs = 500
epochs = 100
dropout = 0.2
```

```python
def my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout):
    if architecture == 'nn':
        model = Sequential()
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(hidden_dims, input_dim=input_shape,activation = 'relu'))
        model.add(Dropout(dropout))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'cnn':
        model = Sequential()
        model.add(Reshape((input_shape, 1),input_shape=(input_shape,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        model.add(MaxPooling1D())
        model.add(Flatten())
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'lstm':
        model = Sequential()
        model.add(Reshape((input_shape, 1), input_shape=(input_shape,)))
        model.add(LSTM(hidden_dims))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'cnn_lstm':
        model = Sequential()
        model.add(Reshape((input_shape, 1),input_shape=(input_shape,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        # model.add(MaxPooling1D())
        model.add(LSTM(lstm_num))
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    elif architecture == 'lstm_cnn':
        model = Sequential()
        model.add(Reshape((input_shape, 1), input_shape=(input_shape,)))
        model.add(LSTM(hidden_dims))
        model.add(Reshape((hidden_dims, 1),input_shape=(hidden_dims,)))
        model.add(Conv1D(filters, kernel_size, padding='valid', activation='relu', strides=1))
        model.add(MaxPooling1D())
        model.add(Flatten())
        model.add(Dense(hidden_dims))
        model.add(Dropout(dropout))
        model.add(Activation('relu'))
        model.add(Dense(1))
        model.add(Activation('linear'))

    return model
```

```python
try:
    os.stat(os.path.join(basedir,'MLForecasts','MomRev'))
except:
    os.mkdir(os.path.join(basedir,'MLForecasts','MomRev'))
for i in range(0,len(ModelList)):
    m = ModelList[i]
    architecture,lossfunction,WeightVar,DepVar = m[0],m[1],m[2],m[3]
    print(architecture, lossfunction, WeightVar, DepVar)
    save_path_head = os.path.join(path,'ChartingByMachines','MLForecasts','MomRev'
                                  ,'_'.join([architecture.upper(),lossfunction.upper(),DepVar,WeightVar]))
    try:
        os.stat(save_path_head)
    except:
        os.mkdir(save_path_head)
    N = 30 # we will run the model 30 times and then average them as the final result
    for period in modelperiod:
        MinMonth = int(period.split('-')[0])
        MaxMonth = int(period.split('-')[1])
        FitObservationsFeaturesCSV = data[(data['Month']>=MinMonth)
                                          & (data['Month']<=MaxMonth)][features+[DepVar] + [WeightVar]]
        TestingX = data[features].values
        for n in range(N):
            Training,Validating = train_test_split(FitObservationsFeaturesCSV,test_size=0.3)
            TrainingX,Trainingy,TrainingW = Training[features].values,Training[DepVar].values,Training[WeightVar].values
            ValidatingX,Validatingy,ValidatingW = Validating[features].values,Validating[DepVar].values,Validating[WeightVar].values
            FolderName = period
            v = 'v'+str(n+1)
            print('period:%s, %s' % (period, v))
            try:
                os.stat(os.path.join(save_path_head,FolderName))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName))
            try:
                os.stat(os.path.join(save_path_head,FolderName,v))
            except:
                os.mkdir(os.path.join(save_path_head,FolderName,v))

            tf.keras.backend.clear_session()
            # early stopping
            es = EarlyStopping(monitor='val_mae', mode='min', verbose=0, patience = 5)
            mc = ModelCheckpoint(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'),
                                 monitor='val_mae', mode='min', verbose=0, save_best_only=True)
            input_shape = TrainingX.shape[1]
            model = my_model(architecture,input_shape,filters,kernel_size,lstm_num,hidden_dims,dropout)
            model.compile(loss=lossfunction, optimizer='adam', metrics=['mae'])
            history = model.fit(TrainingX, Trainingy,
                            sample_weight = TrainingW,
                            batch_size=batch_size,
                            epochs=epochs,
                            validation_data=(ValidatingX, Validatingy),
                            verbose = 0,
                            callbacks = [es, mc])

            saved_model = load_model(os.path.join(save_path_head,FolderName,v,FolderName+'.h5'))

            if lossfunction == 'mse':
                loss = pd.DataFrame(columns=['mse','val_mse','mae','val_mae',])
                loss['mse'] = history.history['loss']
                loss['val_mse'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']
            else:
                loss = pd.DataFrame(columns=['mae','val_mae','mae','val_mae',])
                loss['mae'] = history.history['loss']
                loss['val_mae'] = history.history['val_loss']
                loss['mae'] = history.history['mae']
                loss['val_mae'] = history.history['val_mae']

            if 'PredictedValue' in data.columns:
                data = data.drop(columns=['PredictedValue'])

            y = saved_model.predict(TestingX,batch_size=batch_size)

            data = data.assign(PredictedValue = y)
            data[['Month','permno',DepVar,'PredictedValue']].to_csv(os.path.join(save_path_head,FolderName,v,'Forecast.csv'),index=False)
            loss.to_csv(os.path.join(save_path_head,FolderName,v,'LossFunction.csv'),index=False)
```
