```python
import os
import time
import pyautogui

# 配置目录路径
cpp_dir = r"F:\All_Proj"
save_dir = r"F:\cpp_snapshots1"

# 创建保存目录
def ensure_directory(directory):
    if not os.path.exists(directory):
        try:
            os.makedirs(directory)
        except OSError as e:
            print(f"创建目录失败: {e}")
            return False
    return True

# 截图函数
def snapshot_cpp(file_path, idx):
    try:
        # 检查文件是否存在
        if not os.path.exists(file_path):
            print(f"文件不存在: {file_path}")
            return False

        # 打开文件（VSCode 必须是默认程序）
        os.startfile(file_path)
        time.sleep(1.5)  # 增加等待时间以确保文件完全打开

        # 全选代码
        pyautogui.hotkey("ctrl", "a")
        time.sleep(0.3)

        # 触发 Codesnap 截图 (改成你的快捷键)
        pyautogui.hotkey("ctrl", "shift", "p")
        time.sleep(1.5)

        # 生成图片名称
        img_name = f"Set_{idx:03d}_{os.path.basename(file_path).replace('.cpp', '.png')}"
        img_path = os.path.join(save_dir, img_name)

        # 保存图片
        pyautogui.hotkey("ctrl", "s")
        time.sleep(0.5)

        # 输入路径 & 保存
        pyautogui.typewrite(img_path)
        pyautogui.press("enter")
        time.sleep(1)

        print(f"✅ 已保存: {img_path}")
        return True

    except Exception as e:
        print(f"处理文件 {file_path} 时发生错误: {e}")
        return False

# 主执行函数
def main():
    # 确保保存目录存在
    if not ensure_directory(save_dir):
        return

    # 遍历所有 cpp 文件
    idx = 1
    for root, _, files in os.walk(cpp_dir):
        for f in files:
            if f.endswith(".cpp"):
                file_path = os.path.join(root, f)
                if snapshot_cpp(file_path, idx):
                    idx += 1

if __name__ == "__main__":
    main()
```

```python
import os
import time
import pyautogui
import traceback
from datetime import datetime

# 配置目录路径
cpp_dir = r"F:\All_Proj"
save_dir = r"F:\cpp_snapshots1"

# ✅ 确保保存目录存在
os.makedirs(save_dir, exist_ok=True)

def clean_filename(filename):
    invalid_chars = '<>:"/\\|?*'
    for char in invalid_chars:
        filename = filename.replace(char, '_')
    return filename

# === 截图函数 ===
def snapshot_cpp(file_path, idx):
    try:
        if not os.path.exists(file_path):
            print(f"文件不存在: {file_path}")
            return False

        # ✅ 生成 `Set_001` 风格文件名
        set_name = f"Set_{idx:03d}"
        img_dir = os.path.join(save_dir, set_name)
        os.makedirs(img_dir, exist_ok=True)

        base_name = os.path.basename(file_path).replace(".cpp", "")
        img_name = clean_filename(f"{base_name}.png")
        img_path = os.path.join(img_dir, img_name)

        # 打开文件（VSCode 必须是默认程序）
        os.startfile(file_path)
        time.sleep(2.2)  # 等待 VSCode 打开

        # 全选代码
        pyautogui.hotkey("ctrl", "a")
        time.sleep(0.4)

        # 打开命令面板 → CodeSnap
        pyautogui.hotkey("ctrl", "shift", "p")
        time.sleep(0.5)
        pyautogui.write("CodeSnap")
        time.sleep(0.4)
        pyautogui.press("enter")
        time.sleep(1.5)

        # 保存图片
        pyautogui.hotkey("ctrl", "s")
        time.sleep(1)

        # 清空路径栏 → 输入新路径
        pyautogui.hotkey("ctrl", "a")
        time.sleep(0.3)
        pyautogui.press("delete")
        time.sleep(0.3)

        for char in img_path:
            pyautogui.write(char)
            time.sleep(0.02)

        time.sleep(0.3)
        pyautogui.press("enter")
        time.sleep(1.2)

        # 关闭当前窗口
        pyautogui.hotkey("ctrl", "w")
        time.sleep(0.5)

        # 验证保存是否成功
        if os.path.exists(img_path):
            print(f"✅ 已保存: {img_path}")
            return True
        else:
            print(f"❌ 未找到文件: {img_path}")
            return False

    except Exception as e:
        print(f"处理文件 {file_path} 时发生错误:")
        traceback.print_exc()
        return False

# === 主函数 ===
def main():
    print("开始执行截图任务...")

    cpp_files = []
    for root, _, files in os.walk(cpp_dir):
        for f in files:
            if f.endswith(".cpp"):
                cpp_files.append(os.path.join(root, f))

    total_files = len(cpp_files)
    print(f"找到 {total_files} 个 CPP 文件")

    for idx, file_path in enumerate(cpp_files, start=1):
        print(f"\n处理第 {idx}/{total_files} 个文件: {os.path.basename(file_path)}")
        snapshot_cpp(file_path, idx)
        time.sleep(1)

    print("\n🎉 任务完成!")

if __name__ == "__main__":
    main()
```
